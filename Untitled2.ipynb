```python
from google.colab import files
import cv2, easyocr, glob
from ultralytics import YOLO

# This will open a choose file button right below the cell
uploaded = files.upload()
img_path = list(uploaded.keys())[0]

model, reader = YOLO("best.pt"), easyocr.Reader(['en'])
img = cv2.imread(img_path)

for r in model(img_path):
    for box in r.boxes:
        if model.names[int(box.cls[0])] == "vote_tally":
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            print(reader.readtext(img[y1:y2, x1:x2], detail=0))
```

```python
results = model(img_path)
for r in results:
    for box in r.boxes:
      print(
          f"Found: {model.names[int(box.cls[0])]}"
          f"{float(box.conf[0]):.2f}"
      )
```

```python
import cv2
import easyocr
import re
import pandas as pd

reader = easyocr.Reader(['en'])

img = cv2.imread("IMG_2685.jpeg")
h, w, _ = img.shape

crop = img[int(h * 0.15) : int(h * 0.55)]

res = reader.readtext(crop, detail=0)

data = []

for item in res:
  text = item.strip()
  clean_num = re.sub(r'[^\d]', '', text)

  if clean_num and len(clean_num) <= 5 and not re.search(r'[a-zA-Z]{3,}', text):
    if data:
      data[-1]["votes"] = clean_num
    else:
      if len(text) > 3 and "Candidate" not in text:
        data.append({"candidate": text, "votes": "0"})

df = pd.DataFrame(data)
print(df)

```

```python
import cv2
import easyocr
import pandas as pd

reader = easyocr.Reader(['en'])
img = cv2.imread("IMG_2685.jpeg")
h, w, _ = img.shape

crop = img[int(h * 0.15) : int(h * 0.55)]

# Clean up image contrast
gray = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)
thresh = cv2.threshold(gray, 180, 255, cv2. THRESH_BINARY)[1]

res = reader.readtext(thresh, detail=0)

data = []
for item in res:
  text = item.strip()
  if len(text) > 1:
    data.append({'detected_text': text, 'votes': ''})

df = pd.DataFrame(data)
print(df)
```
